# Day 2: Batch vs. Online Learning

Notes from today's session into how machine learning models handle data pipelines and ingest information in real-world systems.

## 1. Batch Learning (Offline Learning)

By this term, I understood that models are trained on the entire, massive dataset all at once before the model is deployed to the server.

*   **The Problem with Static Logic:** Models trained with this technique don't adapt on their own when new, unseen data patterns arrive. Instead, their performance slowly degrades over time because they don't evolve with incoming datasets once live.
*   **The Cost:** To update a batch system, we have to take it offline, mix the old data with the new data, retrain the entire thing from scratch, and re-deploy it. This cycle is incredibly time-consuming and computationally expensive.
*   **Best for:** Systems where trends change slowly and you don't need real-time adaptations.


## 2. Online Machine Learning (Incremental Learning)

This is the opposite approach. While we still give the model a basic baseline training on historical data before launching it, it doesn't stop learning after deployment. It continuously updates itself as new data enters the system.

*   **How it Works:** Instead of processing massive bulk files, it gets trained using smaller chunks of data (or single data points) sequentially. It understands the shifting nature of the problem and improves itself over time.
*   **Resource Efficiency:** It is highly cost-effective and saves massive storage space. Because the system handles data in smaller portions on the fly, it requires significantly less machine memory—meaning you don't need to retain massive datasets forever.

### The Risks & Disadvantages:
Online ML is much trickier to manage in production because its live behavior depends entirely on the incoming data quality:
1.  **Learning Rate Dependencies:** If the learning rate (the speed at which it learns new things) is set too high and the data updates too quickly, the model can rapidly "forget" the previous important things it learned.
2.  **Data Quality:** If the live streaming data becomes noisy, corrupted, or full of errors, the model will ingest those mistakes instantly. This can cause the system to give completely wrong insights rather than getting better. Continuous monitoring is mandatory.

## Instance-Based vs. Model-Based Learning

This part is all about how the machine actually makes a prediction once a new, unseen input arrives.

### 1. Instance-Based Learning
This basically means the machine just memorizes the training data as-is. It doesn't try to find a deep underlying formula or a general trendline.
*   **How it works:** When new data comes in, the system looks at all its stored past examples, finds the closest match using a similarity check, and copies that outcome.
*   **The Catch:** There is no separate "model file" saved. Because of that, you *have* to keep your raw training data sitting in memory forever to process future inputs. It takes up a lot of storage space during prediction time.
*   **Example:** K-Nearest Neighbors (KNN).

### 2. Model-Based Learning (Finding the Formula)
This is a lot different. Instead of memorizing individual examples, the machine scans the dataset to figure out the underlying patterns, trends, and math formulas.
*   **How it works:** It builds a generalized rule or equation based on what it learned from the past data. When a new problem comes in, it just runs it through that equation.
*   Once the machine builds the formula and saves the model parameters, you can literally throw the raw training data away. This uses way less storage space when you deploy it because you only store the final model, not the massive raw data.
*   **Example:** Linear Regression.
