# Day 3
#Challenges in Machine Learning

Key Learnings

* Data quality is one of the most important factors in ML performance
* Overfitting happens when the model memorizes training data
* Underfitting happens when the model is too simple
* Feature selection improves accuracy and reduces noise
* Training models can be computationally expensive
* Some models are hard to interpret (lack of explainability)


Better data and balanced models matter more than complex algorithms.

#Applications of Machine Learning

### Key Learnings

* Healthcare: disease prediction and medical image analysis
* Finance: fraud detection and risk analysis
* E-commerce: recommendation systems
* NLP: chatbots, translation, and text processing
* Computer Vision: object and face detection
* Autonomous systems: self-driving cars and robotics


Machine learning is used in almost every major industry today.

## Machine Learning Development Life Cycle (MLDLC)

### Key Learnings

* Problem definition is the first and most important step
* Data collection from relevant sources
* Data cleaning and preprocessing
* Feature engineering to improve model performance
* Model training using algorithms
* Model evaluation using performance metrics
* Deployment into real-world applications
* Continuous monitoring and updates

ML is a full pipeline from problem to production, not just model building.

---

## Tensors

### Key Learnings

* Scalar: single value (0D tensor)
* Vector: 1D array of values
* Matrix: 2D grid of numbers
* Tensor: multi-dimensional data structure
* Tensor operations include addition, multiplication, reshaping
* Used in deep learning frameworks like TensorFlow and PyTorch

Tensors are the basic data structure behind deep learning.

---

## Introduction to Data Handling (Bonus Learning)

### Key Learnings

* Understanding datasets is the first step in ML
* Data can be structured (tables) or unstructured (images, text)
* Pandas is used for data manipulation and analysis
* Handling missing values is important before training models
* Data visualization helps understand patterns
* Clean data improves model performance significantly

Good data handling is the foundation of every successful ML project.
